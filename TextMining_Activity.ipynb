## IT414 – Text Mining Laboratory Activity
**Dataset:** Fin_labExer1_dataset.pdf  
**Assigned Pages:** Page 5 (index 4) and Page 35 (index 34)  
**Libraries:** pdfplumber, nltk

---
#### Step 1 – Extract Text Data from the Assigned Pages

```python
import pdfplumber
import nltk
import re
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

PDF_PATH = "Fin_labExer1_dataset.pdf"

with pdfplumber.open(PDF_PATH) as pdf:
    text_page5 = pdf.pages[4].extract_text()
    print(f"Page 5  extracted — Characters: {len(text_page5) if text_page5 else 0}")

    text_page35 = pdf.pages[34].extract_text()
    print(f"Page 35 extracted — Characters: {len(text_page35) if text_page35 else 0}")

combined_text = (text_page5 or "") + "\n" + (text_page35 or "")
print(f"\nCombined text length: {len(combined_text)} characters")
```

---
#### Step 2 – Display Sample of Extracted Raw Text

```python
print("=" * 60)
print("SAMPLE RAW TEXT ── Page 5 (first 500 characters):")
print("=" * 60)
print(text_page5[:500])

print()
print("=" * 60)
print("SAMPLE RAW TEXT ── Page 35 (first 500 characters):")
print("=" * 60)
print(text_page35[:500])
```

---
#### Step 3 – Word-Level Tokenization using word_tokenize()

```python
word_tokens = word_tokenize(combined_text)

print("=" * 60)
print("SAMPLE WORD TOKENS (first 30 tokens):")
print("=" * 60)
print(word_tokens[:30])
```

---
#### Step 4 – Sentence-Level Tokenization using sent_tokenize()

```python
sent_tokens = sent_tokenize(combined_text)

print("=" * 60)
print("SAMPLE SENTENCES (first 5 sentences):")
print("=" * 60)
for i, sentence in enumerate(sent_tokens[:5], start=1):
    print(f"\nSentence {i}:")
    print(sentence)
```

---
#### Step 5 – Total Number of Words

```python
total_words = len(word_tokens)

print("=" * 60)
print("WORD COUNT")
print("=" * 60)
print(f"Total number of words: {total_words}")
```

---
#### Step 6 – Total Number of Sentences

```python
total_sentences = len(sent_tokens)

print("=" * 60)
print("SENTENCE COUNT")
print("=" * 60)
print(f"Total number of sentences: {total_sentences}")
```

---
#### Step 7 – Total Number of Paragraphs

```python
paragraphs = [p.strip() for p in combined_text.split('\n') if p.strip()]
total_paragraphs = len(paragraphs)

print("=" * 60)
print("PARAGRAPH COUNT")
print("=" * 60)
print(f"Total number of paragraphs: {total_paragraphs}")
```

---
#### Summary of Results

```python
print("=" * 60)
print("RESULTS SUMMARY")
print("=" * 60)
print(f"  Pages Analyzed        : Page 5 (index 4) & Page 35 (index 34)")
print(f"  Total Characters      : {len(combined_text)}")
print(f"  Total Words           : {total_words}")
print(f"  Total Sentences       : {total_sentences}")
print(f"  Total Paragraphs      : {total_paragraphs}")
print("=" * 60)
```
